```python
import ipywidgets as widgets
from ipywidgets import interact, interactive
from IPython.display import display

special_groupings = [
    "operation_types",
    "posix_v_stdio"
]

def args(workdir, exp_type, exp_name, limit_runs, operations):
    return (workdir, exp_type, exp_name, limit_runs, operations)

arguments=interactive(args, workdir="../../data/exps", exp_type=["weak_scaling", "strong_scaling"], exp_name=["general", "phases", "sharedfs"], limit_runs=True, continuous_update=False, operations=["read", "write"])
display(arguments)
```

```python
def sample_df(df):
    slimmed_df = df.sample(frac=0.1)
    # print(slimmed_df)
    # slimmed_df = df
    return slimmed_df

# load data
# print(arguments.result)
darshan_data_frames = load_dataframe_dxt(arguments.result[0], arguments.result[1], arguments.result[2], arguments.result[3], dxt_operations=["read", "write"], analysis_func=sample_df)
# print(darshan_data_frames)
# darshan_data_frames.to_parquet("dxt_chunks/darshan_data_frames.parquet", name_function=lambda x: f"dxt_{x}.parquet")
darshan_data_frames
```

```python
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt


color_map = {
    "apptainer": "#45B6EB",
    "none": "#E7BB42",
    "podman": "#62B35F",
    "shifter": "#FF33A8"
}

SCALING=' '.join(arguments.result[1].split('_'))
nprocs=32
operation="write"


shared_writer_starts = darshan_data_frames[(darshan_data_frames["file_name"].str.endswith("DAOD_PHYSLITE.pool.root.1")) & (darshan_data_frames["worker"] == "shared_writer") & (darshan_data_frames["option"] != "250")].groupby(["nprocs", "container", "option"]).agg({"start_time": "min"}).reset_index() #sort_values(["nprocs", "container", "option", "start_time"]).reset_index() #
shared_writer_starts.rename(columns={"start_time": "learning_phase_begins"}, inplace=True)

fig,axes = plt.subplots(nrows=2, ncols=3, figsize=(20, 6), sharex=False, sharey=False)
for idx,option in enumerate(["100", "500", "1000"]):
    # option_df = shared_writer_starts[shared_writer_starts["option"] == option]
    option_subset = darshan_data_frames[(darshan_data_frames["option"] == option) & (darshan_data_frames["nprocs"] == nprocs) & (darshan_data_frames["operation"] == operation)]
    option_subset["length (KB)"] = option_subset["length"] / 1024
    # shared_writer_starts[shared_writer_starts["option"] == option]
    # print(shared_writer_starts[shared_writer_starts["option"] == option])
    plot_df = pd.merge(option_subset, shared_writer_starts, on=["nprocs", "container", "option"], how="left")
    # print(plot_df)
    plot1 = sb.histplot(data=plot_df, x="start_time", hue="container", ax=axes[0][idx], label=option, color=color_map)
    plot1.set_title(f"{operation} operation distribution | auto-flush={option} | nprocs={nprocs}")

    plot2 = sb.histplot(data=plot_df, x="length (KB)", hue="container", multiple="dodge", bins=10, ax=axes[1][idx], log_scale=True, label=option, shrink=0.8, color=color_map)
    
    # print(plot_df)
    # print(plot_df.groupby("container").agg({"learning_phase_begins": "min"}).reset_index())
    for index, row in plot_df.groupby("container").agg({"learning_phase_begins": "min"}).reset_index().iterrows():
        # print(row)
        axes[0][idx].axvline(x=row["learning_phase_begins"], color=color_map[row["container"]], linestyle='--')
    # for sub_idx,start_time in enumerate(plot_df["learning_phase_begins"].unique()):
    #     # print(plot_d].unique())
    #     axes[0][idx].axvline(x=start_time, color='red', linestyle='--')

fig.suptitle(f"Auto-flush Configuration Comparison ({SCALING})")
# shared_writer_starts
# fig
```
