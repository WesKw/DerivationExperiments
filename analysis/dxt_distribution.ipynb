```python
import matplotlib.pyplot as plt
import seaborn as sb

# plot the write start_time distribution
read_frames = darshan_data_frames[darshan_data_frames["operation"] == "read"]
read_frames = read_frames[read_frames["nprocs"] == 16]
read_frames["length (KB)"] = read_frames["length"] / 1024

# write_frames = ...
sb.set_theme(style="whitegrid")
plot = sb.histplot(data=read_frames, x="start_time", hue="container", bins='auto', kde=True)
plot.set_title("Distribution of Reads for 16 Procs")
plt.show()

read_sizes = sb.histplot(data=read_frames, x="length (KB)", hue="container", bins=10, log_scale=True, multiple="dodge", shrink=0.8)
read_sizes.set_title("Distribution of Read sizes for 16 Procs")
plt.show()
```
